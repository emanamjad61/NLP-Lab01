# Task 04 — Sets for Unique Vocabulary

*Topics Covered: Set (`set`) - Core Python Data Structure*

**Part (a)** — Create a list called `raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]`. Convert it into a set called `vocab`. Print both `raw` and `vocab`, and print the number of unique words. Explain in a comment why sets remove duplicates automatically.

```python
# Task 04(a)  |  Eman Amjad  |  FA23-BAI-013
raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]
vocab = set(raw)

print(f"raw   ({len(raw)} items) : {raw}")
print(f"vocab ({len(vocab)} items) : {vocab}")
print(f"\nDuplicates dropped: {len(raw) - len(vocab)}")

# Why do duplicates disappear on their own?
# Every item added to a set is first run through hash(). The hash decides which
# slot the item belongs to, and Python checks that slot before storing anything.
# If an equal item is already sitting there, the new one is thrown away instead
# of being added. That check is built into the data structure, so a set simply
# cannot hold the same value twice.
```

```python
# Task 06(d)  |  Eman Amjad  |  FA23-BAI-013
def normalize(text):
    """Trim, lowercase, then drop every comma and full-stop."""
    result = text.strip().lower()
    for mark in [",", "."]:
        result = result.replace(mark, "")
    return result


messy = [
    "  TOKENIZATION, stemming, and lemmatisation.  ",
    "   Text Mining. Information Retrieval.   ",
    "  CORPUS, vocabulary, embeddings.  "
]

for number, item in enumerate(messy, start=1):
    print(f"Test {number}")
    print(f"   before -> [{item}]")
    print(f"   after  -> [{normalize(item)}]")
```

```python
# Task 07(c)  |  Eman Amjad  |  FA23-BAI-013
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words, pure_numbers, mixed_tokens = [], [], []

for item in words:
    if item.isalpha():
        pure_words.append(item)
    elif item.isdigit():
        pure_numbers.append(item)
    else:
        mixed_tokens.append(item)

print(f"Input         : {words}\n")
print(f"pure_words    ({len(pure_words)}): {pure_words}")
print(f"pure_numbers  ({len(pure_numbers)}): {pure_numbers}")
print(f"mixed_tokens  ({len(mixed_tokens)}): {mixed_tokens}")
```

**Part (d)** — Write a function called `validate_token(token)` that returns the string "WORD" if the token is purely alphabetic, "NUMBER" if it is purely numeric, and "MIXED" otherwise. Apply this function to each element in the list `["chat", "99", "gpt4", "bert", "2024", "nlp!"]` using a loop and print the token alongside its label.

```python
# Task 07(d)  |  Eman Amjad  |  FA23-BAI-013
def validate_token(token):
    """Label a token as WORD, NUMBER or MIXED."""
    if token.isalpha():
        return "WORD"
    if token.isdigit():
        return "NUMBER"
    return "MIXED"


sample = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]
labels = {}

for token in sample:
    label = validate_token(token)
    labels[token] = label
    print(f"   {token:<8} -> {label}")

print(f"\nCollected as a dictionary: {labels}")
```

```python
# Task 08(a)  |  Eman Amjad  |  FA23-BAI-013
def clean(text):
    """Trim the edges and lowercase the whole string."""
    return text.strip().lower()


clean_lambda = lambda text: text.strip().lower()

demo = "  HELLO Natural Language Processing  "

print(f"Input          : [{demo}]")
print(f"clean()        : [{clean(demo)}]")
print(f"clean_lambda() : [{clean_lambda(demo)}]")
print(f"\nSame output from both? {clean(demo) == clean_lambda(demo)}")

# A lambda is just a compact way to write a small function that fits on one line
```

**Part (b)** — Write a function called `tokenize(text)` that takes a cleaned string and returns a list of individual word tokens using `split()`. Then write another function called `remove_short(tokens, n)` that takes the token list and an integer n, and returns only tokens whose length is greater than n. Chain both functions on the sentence "I love working with NLP tools in Python daily" with n = 3 and print the result.

```python
# Task 08(b)  |  Eman Amjad  |  FA23-BAI-013
def tokenize(text):
    """Break a cleaned string into word tokens."""
    return text.split()


def remove_short(tokens, n):
    """Return only those tokens longer than n characters."""
    kept = []
    for token in tokens:
        if len(token) > n:
            kept.append(token)
    return kept


raw_sentence = "I love working with NLP tools in Python daily"

step_tokens = tokenize(clean(raw_sentence))
print(f"Tokenised          : {step_tokens}")

step_filtered = remove_short(step_tokens, 3)
print(f"Longer than 3 chars: {step_filtered}")
print(f"\nDropped {len(step_tokens) - len(step_filtered)} short tokens.")
```

```python
# Task 08(c)  |  Eman Amjad  |  FA23-BAI-013
def word_lengths(tokens):
    """Build a word -> length dictionary one entry at a time."""
    lengths = {}
    for token in tokens:
        lengths[token] = len(token)
    return lengths


sample_tokens = ["language", "model", "ai", "processing", "deep"]
result = word_lengths(sample_tokens)

print(f"Tokens     : {sample_tokens}")
print(f"Dictionary : {result}")

longest = max(result, key=result.get)
print(f"\nLongest word: '{longest}' with {result[longest]} characters")
```

**Part (a)** — Given `words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]`. Use `map()` with a lambda to: (i) convert every word to lowercase - print the result as a list, and (ii) compute the character length of each word - print the result as a list.

```python
# Task 09(a)  |  Eman Amjad  |  FA23-BAI-013
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lowercased = list(map(lambda item: item.lower(), words))
sizes = list(map(lambda item: len(item), words))

print(f"Original   : {words}")
print(f"(i)  lower : {lowercased}")
print(f"(ii) length: {sizes}")

# Pairing the two results back together makes them easier to read
print(f"\nWord/length pairs: {list(zip(words, sizes))}")
```

```python
# Task 09(b)  |  Eman Amjad  |  FA23-BAI-013
words_long = list(filter(lambda item: len(item) > 3, words))
words_with_t = list(filter(lambda item: item.lower().count("t") > 0, words))

print(f"All words              : {words}\n")
print(f"(i)  more than 3 chars : {words_long}")
print(f"(ii) containing a 't'  : {words_with_t}")

# 'Python' and 'text' both qualify for the second list, and the .lower()
# inside the lambda is what lets a capital T match as well.
```

**Part (d)** — Import `Counter` from `collections`. Given the sentence "nlp nlp python ai nlp python text ai nlp python text", split it into words and create a Counter called `freq`. Print the full counter, then print: (i) the 3 most common words using `most_common()`, (ii) the frequency of the word "python", and (iii) the total number of words and the size of the unique vocabulary.

```python
# Task 09(d)  |  Eman Amjad  |  FA23-BAI-013
from collections import Counter

corpus = "nlp nlp python ai nlp python text ai nlp python text"
word_list = corpus.split()
freq = Counter(word_list)

print(f"Corpus  : {corpus}")
print(f"Counter : {freq}\n")

print("(i)   Top 3 words:")
for rank, (word, count) in enumerate(freq.most_common(3), start=1):
    print(f"        {rank}. {word} appears {count} times")

print(f"\n(ii)  'python' occurs {freq['python']} times")

print(f"\n(iii) Total words in the corpus : {sum(freq.values())}")
print(f"      Unique vocabulary size    : {len(freq)}")
```
